## Importing and Inspecting the Data

I'm going to import the insurance.csv file and check the number
of records, the column names, and the first few records.

```python
import csv

data = []

with open("insurance.csv") as archive:
    reader = csv.DictReader(archive)

    for line in reader:
        data.append(line)
```

```python
print("Number of records:", len(data))
```

```python
for person in data[:5]:
    print(person)
```

## Initial Inspection

I imported the insurance.csv file and examined the first
five records.

The columns are: age, sex, bmi, children, smoker, region,
and charges.

Each record was stored as a dictionary within
the data list. The values were imported as text.

## Data Quality Checks

Before converting the numeric columns, I will check for empty
fields and repeated records.

```python
empty_fields = 0

for person in data:
    for column in person:
        if person[column].strip() == "":
            empty_fields = empty_fields + 1

print("Number of empty fields:", empty_fields)
```

```python
seen_records = []
repeated_records = 0

for person in data:
    if person in seen_records:
        repeated_records = repeated_records + 1
    else:
        seen_records.append(person)

print("Number of repeated records:", repeated_records)
```

The dataset contains 1,338 records and no empty fields.

One record repeats all the values of a previous record.
I kept both records because the dataset does not contain
an individual identifier. Identical values alone do not
prove that the records represent the same person.

## Organizing the Data

I will store each column in a separate list.
Numeric values will be converted to numbers,
so they can be used in calculations.

```python
age = []
sex = []
bmi = []
children = []
smoker = []
region = []
charges = []

for person in data:
    age.append(int(person["age"]))
    sex.append(person["sex"])
    bmi.append(float(person["bmi"]))
    children.append(int(person["children"]))
    smoker.append(person["smoker"])
    region.append(person["region"])
    charges.append(float(person["charges"]))
```

```python
#checking
print("Minimum age:", min(age))
print("Maximum age:", max(age))

print("Minimum BMI:", min(bmi))
print("Maximum BMI:", max(bmi))

print("Minimum number of dependents:", min(children))
print("Maximum number of dependents:", max(children))

print("Minimum charges:", min(charges))
print("Maximum charges:", max(charges))

print("First five ages:", age[:5])
print("First five regions:", region[:5])
print("First five charges:", charges[:5])

print("Number of ages:", len(age))
print("Number of charges:", len(charges))
```

```python
# Check the category values before starting the analysis.
unexpected_categories = 0

for current_sex in sex:
    if current_sex != "female" and current_sex != "male":
        print("Unexpected sex category:", current_sex)
        unexpected_categories = unexpected_categories + 1

for current_status in smoker:
    if current_status != "yes" and current_status != "no":
        print("Unexpected smoking category:", current_status)
        unexpected_categories = unexpected_categories + 1

expected_regions = ["northeast", "northwest", "southeast", "southwest"]

for current_region in region:
    if current_region not in expected_regions:
        print("Unexpected region category:", current_region)
        unexpected_categories = unexpected_categories + 1

print("Number of unexpected category values:", unexpected_categories)
```

The ages range from 18 to 64 years.

The categorical values match the expected categories.
All numeric columns were successfully converted to numbers.

The age range limits the analysis: the dataset does not
include children or adults older than 64.

## Data Analysis

I will create functions to calculate averages and compare
medical costs across different groups.

```python
def calculate_average(values):
    total = 0

    for value in values:
        total = total + value

    average = total / len(values)

    return average
```

```python
average_age = calculate_average(age)
average_charges = calculate_average(charges)

print("Average age:", round(average_age, 2))
print("Average charges:", round(average_charges, 2))
```

```python
from statistics import median

median_charges = median(charges)

print("Median costs:", round(median_charges, 2))
```

## General Results

The average age is 39.21 years.

Average medical charges are USD 13,270.42, while the median is USD 9,382.03.

The mean is higher than the median, suggesting that some
high charges are pulling the average upward. For this reason,
both measures are useful when describing the costs.

```python
def filter_charges(categories, charges, group):
    group_charges = []

    for position in range(len(charges)):
        if categories[position] == group:
            group_charges.append(charges[position])

    return group_charges
```

```python
smoker_charges = filter_charges(smoker, charges, "yes")
non_smoker_charges = filter_charges(smoker, charges, "no")
```

```python
average_smoker_charges = calculate_average(smoker_charges)
average_non_smoker_charges = calculate_average(non_smoker_charges)

print("Number of smokers:", len(smoker_charges))
print("Average smoker charges:", round(average_smoker_charges, 2))
print("Median smoker charges:", round(median(smoker_charges), 2))

print()

print("Number of non-smokers:", len(non_smoker_charges))
print("Average non-smoker charges:", round(average_non_smoker_charges, 2))
print("Median non-smoker charges:", round(median(non_smoker_charges), 2))
```

## Smoking Status Results

The dataset contains 274 smokers and 1,064 non-smokers.

Average charges are USD 32,050.23 for smokers and USD 8,434.27
for non-smokers. Smokers have approximately 3.8 times
the average charges of non-smokers.

The median is also higher for smokers: USD 34,456.35,
compared with USD 7,345.41 for non-smokers.

Both measures show higher charges among smokers in this sample.

```python
region_names = ["northeast", "northwest", "southeast", "southwest"]

for current_region in region_names:
    region_charges = filter_charges(region, charges, current_region)

    average_region_charges = calculate_average(region_charges)
    median_region_charges = median(region_charges)

    print("Region:", current_region)
    print("Number of people:", len(region_charges))
    print("Average charges:", round(average_region_charges, 2))
    print("Median charges:", round(median_region_charges, 2))
    print()
```

## Regional Results

The southeast has the highest average charges at USD 14,735.41.
The southwest has the lowest average at USD 12,346.94.

However, the northeast has the highest median at USD 10,057.65.
The southwest has the lowest median at USD 8,798.59.

The regional ranking therefore depends on whether the mean
or median is used.

Differences in age, BMI, and smoking status between regions
may also be related to these results.

## Medical Charges by Age Group

I will compare charges across three age groups:
18–29, 30–49, and 50–64.

These groups were chosen to organize the comparison.

```python
age_groups = []

for current_age in age:
    if current_age < 30:
        age_groups.append("18-29")
    elif current_age < 50:
        age_groups.append("30-49")
    else:
        age_groups.append("50-64")
```

```python
age_group_names = ["18-29", "30-49", "50-64"]

for current_group in age_group_names:
    group_charges = filter_charges(age_groups, charges, current_group)

    average_group_charges = calculate_average(group_charges)
    median_group_charges = median(group_charges)

    print("Age group:", current_group)
    print("Number of people:", len(group_charges))
    print("Average charges:", round(average_group_charges, 2))
    print("Median charges:", round(median_group_charges, 2))
    print()
```

Average charges increase across the selected age groups:
USD 9,182.49 for ages 18–29, USD 13,123.59 for ages 30–49,
and USD 17,902.55 for ages 50–64.

Median charges also increase across these groups.

In this sample, older age groups have higher medical charges.
This comparison does not account for all other differences
between the groups.

## Smoking Status Within Age Groups

I will compare smokers and non-smokers within each age group
to investigate whether the difference appears across
different age ranges.

```python
smoking_statuses = ["yes", "no"]

for current_group in age_group_names:
    print("Age group:", current_group)

    for current_status in smoking_statuses:
        selected_charges = []

        for position in range(len(charges)):
            if age_groups[position] == current_group and smoker[position] == current_status:
                selected_charges.append(charges[position])

        average_selected_charges = calculate_average(selected_charges)

        print("Smoker:", current_status)
        print("Number of people:", len(selected_charges))
        print("Average charges:", round(average_selected_charges, 2))
        print("Median charges:", round(median(selected_charges), 2))
        print()
```

Smokers have higher average charges than non-smokers
in all three age groups.

Among people aged 18–29, average charges are USD 27,518.04
for smokers and USD 4,418.57 for non-smokers.

Among people aged 50–64, the averages are USD 38,748.35
for smokers and USD 13,430.90 for non-smokers.

The difference between smokers and non-smokers therefore
appears within each selected age group.

However, these groups may still differ in BMI, exact age,
and other characteristics.

Smoking status appears to be a useful candidate for predicting medical
charges. Smokers had higher average charges overall and within every
selected age group.

Age may also be useful because both mean and median charges increased
across the selected age groups.

These findings suggest that smoking status and age deserve further
investigation in a predictive model.

However, this analysis did not measure the independent importance of
each feature or test prediction accuracy. BMI and other variables would
also need to be investigated before ranking the features.


## Potential Bias and Use Cases

### Age Coverage

The sample includes people aged 18 to 64. Children and adults older
than 64 are not represented. The results should not automatically
be applied to these groups.

### Geographic Coverage

The dataset describes individuals from four U.S. regions. The findings
may not apply to countries with different healthcare systems.

### Group Sizes

There are 1,064 non-smokers and 274 smokers in the sample. Smaller
subgroups provide less information for comparisons, and their estimates
may be less precise. Unequal group sizes alone do not prove that the
dataset is biased.

### Sampling and Representativeness

This analysis does not establish how individuals were selected for
the dataset. It therefore cannot confirm that the sample represents
the wider population of insurance beneficiaries.

### Missing Information

The available columns do not describe factors such as income, medical
conditions, or details of insurance coverage. These factors could help
explain differences in charges. Their absence limits the interpretation
of the observed associations.

### Potential Uses

The dataset is useful for practicing exploratory analysis and developing
hypotheses for future prediction work.

This analysis does not establish that the findings can accurately
predict an individual's medical charges. Any predictive application
would require model evaluation and checks on the population where
it would be used.
